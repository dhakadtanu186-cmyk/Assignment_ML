```python
# Select the Name, Hours_Studied, and Marks columns. Then identify students whose marks are above the dataset's average.
selected_data=df[['Name','Hours_Studied','Marks']]
print('Sc',selected_data)

average_marks=df['Marks'].mean()
print("Average_marks=",average_marks)

above_average=selected_data[selected_data['Marks']> average_marks]
print(above_average)
```

```python
# 1. Reset the index after filtering, compare the original and new indices, and explain why an index should not automatically be treated as a student's unique identifier.
print("Original Index:")
print(above_average)

new_index = above_average.reset_index(drop=True)

print("\nAfter Reset Index:")
print(new_index)
```

```python
# Select rows at positions 2 through 7 using iloc, returning only Name, Attendance, and Marks. Explain why the ending position behaves differently from the equivalent label-based selection.
result=df.iloc[2:8][["Name","Attendance","Marks"]]
print(result)
```

```python
# Create a copy of the DataFrame and set Student_ID as its index. Use loc to retrieve students with IDs 103, 107, and 110.
df_copy=df.copy() 
df_copy = df_copy.set_index('Student_ID')
result = df_copy.loc[[103, 107, 110]]
print("Selected Students:")
print(result)
```

```python
# Using Boolean conditions, identify students who studied at least 6 hours, have attendance of at least 85%, and scored more than 80 marks.
result = df[
    (df['Hours_Studied'] >= 6) &
    (df['Attendance'] >= 85) &
    (df['Marks'] > 80)
]

print("Students satisfying all conditions:")
print(result)
```

```python
# Using iloc, extract the last four students and the last three columns without writing any column names explicitly.
result = df.iloc[-4:, -3:]

print("Last four students and last three columns:")
print(result)
```

```python
# Create a new DataFrame containing students whose marks are below the median. Preserve their original student IDs and explain why filtering rows and resetting the index can produce different index labels.
median_marks=df['Marks'].median()
print("Median_Marks:",median_marks)
below_median=df[df['Marks']<median_marks].copy()

print("\n Students scoring below median:")
print(below_median)
```
